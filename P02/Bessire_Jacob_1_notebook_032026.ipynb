### *Identification et suppression des lignes "non pays" dans le dataframe dfcountry*

En regardant un peu dfcountry on voit qu'il y a des lignes qui correspondent à des regions du monde ou autres entités qui regroupent plusieurs pays.  
On peut identifier facilement ces lignes ou les colonnes 'Regions' ou 'Currency Unit' sont vides (seul un ou 2 'vrais pays' sont également dans le cas)  
Pour supprimer les lignes on recuperer les lignes dont le champ 'Region' est null/NaN puis on supprime les lignes en question

```python
# Certaines lignes representent des regions, des groupes, ...  et non pas des pays
themaskregions=(dfcountry['Region'].isna()) & (~dfcountry['Short Name'].isin(["Nauru","Gibraltar"]))   #exclure Nauru et Gibraltar qui sont des pays
dfcountry[themaskregions].iloc[:,:7]
```

```python
# Memorisation des "Country Codes" des lignes <non pays> dans la variable code_nonpays
codes_nonpays=dfcountry[themaskregions]['Country Code'].tolist()

# Suppression des lignes <non pays> de dfcountry
display("Avant:",dfcountry.shape[0])
todrop=dfcountry[themaskregions].index
dfcountry.drop(index=todrop, inplace=True)
display("Apres:",dfcountry.shape[0])
```

```python
codes_nonpays   #les Country codes des lignes supprimés
```

### Creation de la table pivot (dfpivot =  Pays vs Indicateur)

```python
dfdata2.info()
```

```python
# On transforme les colonnes années en une seule colonne 'Value'
annees = [col for col in dfdata2.columns if col.isnumeric()]
dfmelt = pd.melt(dfdata2, 
                  id_vars=['Country Name', 'Indicator Code'], 
                  value_vars=annees, 
                  var_name='Year', 
                  value_name='Value')

# On s'assure que Year est bien numérique pour un tri correct
dfmelt['Year'] = pd.to_numeric(dfmelt['Year'])

# On trie par pays et par année (du plus vieux au plus récent)
dfmelt = dfmelt.sort_values(by=['Country Name', 'Year'])

# On crée le nouveau DataFrame pivot
dfpivot = dfmelt.pivot_table(
    index='Country Name',      # 1 ligne = 1 pays
    columns='Indicator Code',  # 1 col = 1 indicateur
    values='Value',            # La valeur numérique à agréger
    aggfunc='last'             # La méthode d'agrégation (la derniere année renseignée)
)
dfpivot = dfpivot.reset_index()
dfpivot.head(10)
```

### Rendre la table pivot plus lisible (renommage colonnes indicateurs et lisibilité des chiffres)

```python

dictionnaire_indicateurs = {                               # Pour renommer les colonnes
    'IT.NET.USER.P2':       'Internet_Users_Pct',
    'NY.GDP.MKTP.PP.CD':    'PIB_Total_PPA (B)',
    'NY.GDP.PCAP.PP.CD':    'PIB_Habitant_PPA',
    'NY.GNP.PCAP.PP.CD':    'RNI_Habitant_PPA',
    'SE.SEC.ENRL.GC':       'Inscrits_Secondaire (M)',
    'SP.POP.0014.TO.ZS':    'Jeunes_0_14_Pct',
    'SP.POP.1524.TO.UN':    'Jeunes_15_24 (M)',
    'SP.POP.1564.TO.ZS':    'Actifs_15_64_Pct',
    'SP.POP.TOTL':          'Population_Tot (M)',
    'SP.PRM.GRAD.TO':       'Diplomes_Primaire (M)',
    'SP.SEC.TOTL.IN':       'Inscrits_Secondaire_Tot (M)',
    'SP.TER.TOTL.IN':       'Inscrits_Superieur_Tot (M)'
}

dfpivot = dfpivot.rename(columns=dictionnaire_indicateurs)  # Renommage des colonnes

# Transformation des chiffres (en millions ou en milliard) pour lisibilité
dfpivot['Population_Tot (M)']           = dfpivot['Population_Tot (M)']             / 1_000_000 
dfpivot['Jeunes_15_24 (M)']             = dfpivot['Jeunes_15_24 (M)']               / 1_000_000 
dfpivot['Inscrits_Secondaire_Tot (M)']  = dfpivot['Inscrits_Secondaire_Tot (M)']    / 1_000_000 
dfpivot['Inscrits_Secondaire (M)']      = dfpivot['Inscrits_Secondaire (M)']        / 1_000_000 
dfpivot['Diplomes_Primaire (M)']        = dfpivot['Diplomes_Primaire (M)']          / 1_000_000 
dfpivot['Inscrits_Superieur_Tot (M)']   = dfpivot['Inscrits_Superieur_Tot (M)']     / 1_000_000 

dfpivot['PIB_Total_PPA (B)']            = dfpivot['PIB_Total_PPA (B)']              / 1_000_000_000 

dfpivot = dfpivot.round(2)                                  # tout arrondir (lisibilité)

dfpivot.head(50)
```

### Scoring des pays (methode quantitative: RobustScaler/MinMaxScaler + ponderation des indicateurs)

```python
from sklearn.preprocessing import RobustScaler, MinMaxScaler

rs = RobustScaler()    # RobustScaler pour les indicateurs avec outliers
mms = MinMaxScaler()   # Traitement pour les autres indicateurs 

dfpivot5 = dfpivot4.rename(columns=lambda x: x.replace('_Pct', '').replace(' (B)', '').replace(' (M)', ''))

dfpivot5[['PIB_Total_PPA', 'Jeunes_15_24']] = rs.fit_transform(dfpivot5[['PIB_Total_PPA', 'Jeunes_15_24']])

dfpivot5[['Internet_Users', 'Actifs_15_64']] = mms.fit_transform(dfpivot5[['Internet_Users', 'Actifs_15_64']])

#  remettre tout sur une échelle 0-1 
thecols=['PIB_Total_PPA', 'Jeunes_15_24','Internet_Users', 'Actifs_15_64']
dfpivot5[thecols] = mms.fit_transform(dfpivot5[thecols])

# Ponderation des indicateurs
poids = {
    'Internet_Users': 0.35,         
    'Jeunes_15_24': 0.35,
    'PIB_Total_PPA': 0.15,
    'Actifs_15_64': 0.15
}

# Calcul du score pondéré
dfpivot5['Score_Final'] = (
    dfpivot5['Internet_Users'] * poids['Internet_Users'] +
    dfpivot5['Jeunes_15_24'] * poids['Jeunes_15_24'] +
    dfpivot5['PIB_Total_PPA'] * poids['PIB_Total_PPA'] +
    dfpivot5['Actifs_15_64'] * poids['Actifs_15_64']
) * 100

dfpivot5.sort_values(by='Score_Final', ascending=False).round(2).head(10)
```
